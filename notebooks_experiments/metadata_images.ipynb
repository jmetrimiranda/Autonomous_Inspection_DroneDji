## Notebook Metadata images

```python
from exif import Image

def display_metadata(image_path):
    with open(image_path, 'rb') as image_file:
        img = Image(image_file)

    if img.has_exif:
        print(f"Image contains EXIF metadata (version {img.exif_version})\n")
        print("Metadata Tags:")
        # Iterate over all EXIF tags and print their values
        for tag in img.list_all():
            value = img.get(tag)
            print(f"  {tag}: {value}")
    else:
        print("Image does not contain any EXIF information.")

# Replace 'path/to/your/image.jpg' with the actual path to your image file
display_metadata(r'/home/jorgemetri/Desktop/Autonomous_Inspection_DroneDji/yolo_training/src/Drone_GMU-2/train/images/org_0aec18a68b72702e_1770903010000_jpg.rf.3fa88e5724c06e218d75568e21d3c0fb.jpg')
```
